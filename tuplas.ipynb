## Tuplas
Tuplas são estruturas muito semelhantes às listas. Eles diferem de duas maneiras:

1. Elas usam _parênteses_ em vez de _colchetes_:

```python
tupla_1 = (1, 2, 3)
tupla_2 = 1, 2, 3

print(tupla_1)
print(tupla_2)
```

2. Elas **não** podem ser modificadas (**imutáveis**)

```python
tupla_1[0] = 10
```

#### Convertendo uma lista em tuplas e vice-versa

```python
planetas = ['Mercúrio', 'Vênus', 'Terra', 'Marte', 'Júpiter', 'Saturno', 'Urano', 'Netuno']
print(planetas)

planetas_tupla = tuple(planetas)
print(planetas_tupla)
```

```python
planetas_lista = list(planetas_tupla)
print(planetas_lista)
```

```python
planetas[0] = planetas[0].upper()  # deixa todas as letras do primeiro planeta em maiúscula
print(planetas)
```

```python
planetas_tupla[0] = planetas_tupla[0].upper()
print(planetas_tupla[0])
```

### Descompactando tuplas

```python
notas_do_semestre = (5.0, 10, 2)

prova_1, prova_2, projeto = notas_do_semestre

print(prova_1)
print(prova_2)
print(projeto)
```

### Zip: Compactando listas

```python
carros = ['Ônix', 'Gol', 'Uno']
precos = [50000, 30000, 20000]

carros_com_precos_zip = zip(carros, precos)
carros_com_precos_zip
```

```python
carros_com_precos = list(carros_com_precos_zip)
carros_com_precos
```

```python
carros_com_precos_d = dict(zip(carros, precos))
```

```python
print(carros_com_precos_d)
```

# Iterators



```python
#consideremos os seguintes loops:

for x in [1,2,3,4] :
  print(x**2, end=' ')

print()

for x in [1,2,3,4]:
  print (x**3, end= ' ')

print()

for x in 'ifsp':
  print (x * 2, end= ' ')
```

Em Python, um objeto é considerado *iterável* se ele for ou uma sequência armazenada fisicamente ou se ele produzir um resultado por vez dentro de um processo qualquer que se repita, como um loop. Ou seja, ou são sequências já criadas ou resultados de um algoritmo que pode criá-las ao rodar. Podemos obter iteradores a partir de objetos iteráveis como, por exemplo, listas.

```python
numeros = [10, 20, 30] # uma lista qualquer

it = iter(numeros)  # iter() é uma função que produz um iterador

print(next(it))     # já a função next implementa o protocolo (o mecanismo) da iteração
print(next(it))
print(next(it))
```

Se executarmos uma nova chamada a next, teremos uma sinalização de fim de iteração, porque não existem mais elementos:

```python
next(it)
```

Outros exemplos de iteradores

```python
carros_com_precos_zip = zip(carros, precos)  # vamos refazer o zip
print(next(carros_com_precos_zip))
print(next(carros_com_precos_zip))
```

```python
nomes = ["Alice", "Bob", "Carla"]

e = enumerate(nomes)    # já havíamos encontrado o enumerate anteriormente, que gera um índice

print(next(e))
print(next(e))
```

# map( )

A função map() aplica uma função qualquer aos elementos de um **iterável** e retorna um **iterador** (*iterator*)

Essas funções são geralmente implementadas por funções lambda, definidas como lambda x :

```python
numeros = [2, 3, 4, 5]

m = map(lambda x: x**2, numeros)

print(next(m))
print(next(m))
print(list(m))
```

# filter( )

Também gera iteradores usando funções lambda

```python
numeros = [1, 2, 3, 4, 5, 6]

f = filter(lambda x: x%2 == 0, numeros) # filtra os números pares de uma lista

print(next(f))
print(next(f))
```

# Expressões geradoras (generators)

Semelhante a uma *list comprehension*
Note, porém, que está entre parênteses, não gera uma lista

```python
quadrados = (x ** 2 for x in range(5))
```

```python
print(next(quadrados))
print(next(quadrados))
print(next(quadrados))
```

# yield( )

Permite criar iteradores usando *generators*

```python
def contador():   #criando uma função
  yield 10
  yield 20
  yield 30
```

```python
c = contador()

print(next(c))  # a função fica suspensa, preserva seu estado e continua daquele ponto na próxima chamada
print(next(c))
print(next(c))
```

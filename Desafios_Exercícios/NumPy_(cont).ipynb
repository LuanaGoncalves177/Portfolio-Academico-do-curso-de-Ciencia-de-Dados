# Conceitos básicos sobre arrays NumPy : Remodelação, Concatenação e Separação.

```python
import numpy as np
```

# Remodelação (*reshape*)

Por vezes é conveniente colocar uma sequência de 1 a n números em uma grade m X m

```python
grid = np.arange(1, 10)
print(grid)
```

```python
grid = np.arange(1, 10).reshape(3,3)
print (grid)
```

**Atenção**: para funcionar, o tamanho do array inicial deve ser compatível com tamanho do array remodelado!
O método *reshape* retorna uma view sem cópia do array inicial

```python
grid2 = np.arange(1, 9).reshape(3,3)
```

```python
x = np.array([1,2,3])  #cria um array horizontal
x.reshape((1,3))
```

```python
x.reshape((3,1))       #cria um array vertical
```

É comum que este tipo de conversão de formato seja feito ao mesmo tempo que um fatiamento

```python
x[np.newaxis, :]  #explicita que o eixo é horizontal (escreve como colunas)
```

```python
x[:, np.newaxis]  #explicita que o eixo é vertical (escreve como linhas)
```

# Concatenação

Não necessariamente precisamos manipular vetores um a um, podemos reuní-los em um só vetor maior ou dividir uma grande sequência em vetores que o compõem

```python
x = np.array([1, 2, 3])
y = np.array([3, 2, 1])
np.concatenate([x,y])
```

```python
z = np.array([99, 99, 99])
print(np.concatenate([x,y,z]))
```

```python
matriz = np.array([[1, 2, 3], [4, 5, 6]])
np.concatenate([matriz, matriz])
```

```python
np.concatenate([matriz, matriz], axis=1)  #explicita que a concatenação é horizontal
```

Se for necessário concatenar arrays de diferentes dimensões, por exemplo, vetores unidimensionais com matrizes, pode-se usar as funções *vstack* e *hstack*

```python
np.vstack([x, matriz])
```

```python
y = np.array([[99],[99]])
np.hstack([matriz, y])
```

# Separação

É a operação inversa da concatenação

Notem: N pontos de corte retornarão N+1 subarrays

```python
x = [1, 2, 3, 99, 99, 3, 2, 1]
x1, x2, x3 = np.split(x, [3,5])
print (x1, x2, x3)
```

Analogamente, também, matrizes podem ser separadas

```python
matriz = np.arange(16).reshape((4,4)) #vamos gerar uma matriz 4 x 4
matriz
```

Existem, analogamente, as funções *hsplit* e *vsplit*

```python
superior, inferior = np.vsplit(matriz, [2])
print(superior)
print(inferior)
```

```python
esquerda, direita = np.hsplit(matriz, [2])
print(esquerda)
print(direita)
```

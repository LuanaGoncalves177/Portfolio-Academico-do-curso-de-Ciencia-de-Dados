1) As temperaturas máximas registradas durante uma semana foram

armazenadas em um array NumPy:

import numpy as np

temperaturas = np.array([28, 31, 27, 30, 32, 29, 33])

Utilizando funções de agregação do NumPy:

a) Determine a menor temperatura registrada.

b) Determine a maior temperatura registrada.

c) Calcule a temperatura média da semana.

d) Calcule a soma de todas as temperaturas.

e) Determine em qual posição do array ocorreu a maior temperatura utilizando argmax().


```python
import numpy as np
```

```python
temperaturas = np.array([28, 31, 27, 30, 32, 29, 33])

print(f'a) Menor temperatura registrada: {temperaturas.min()}')
print(f'b) Maior temperatura registrada: {temperaturas.max()}')
print(f'c) Temperatura média da semana: {temperaturas.mean()}')
print(f'd) Soma de todas as temperaturas: {temperaturas.sum()}')
print(f'e) Posição da maior temperatura: {temperaturas.argmax()}')
```

2) Uma loja possui os preços dos seguintes produtos:
precos = np.array([100.0, 250.0, 80.0, 300.0, 150.0])

A loja decidiu conceder 10% de desconto em todos os produtos.

a) Utilizando uma única operação, crie um novo array contendo os preços com 10% de desconto.

b) Some R$ 5,00 a todos os preços do array original.

c) Execute a expressão abaixo e observe seu resultado:
precos > 150. *Explique o significado dos valores True e False obtidos.*

d) Utilize o resultado de uma comparação para mostrar somente os preços maiores que R$ 150,00.

e) Mostre somente os preços entre R$ 100,00 e R$ 250,00, inclusive.

```python
precos = np.array([100.0, 250.0, 80.0, 300.0, 150.0])
precos_com_desconto = precos * 0.9
print(f'a) Preços com 10% de desconto: {precos_com_desconto}')
print(f'b) Preços com 5 reais a mais: {precos +5}')
print(f'c) Preços maiores que 150: {precos > 150}')
print(f'c) Explicação: Os valores menores que 150 receberão o valor False e os maiores o valor True ')
print(f'd) Preços maiores que 150: {precos[precos > 150]}')
print(f'e) Preços entre 100 e 250: {precos[(precos >= 100) & (precos <= 250)]}')
```

3) Considere um array contendo as notas de oito alunos:
notas = np.array([6.5, 8.0, 4.5, 9.0, 7.5, 5.0, 10.0, 7.0])

a) Exiba as notas dos alunos que ocupam as posições 1, 3 e 6 utilizando uma lista de índices: indices = [1, 3, 6]

b) Crie uma nova seleção contendo as notas das posições 7, 2, 0 e 5, nessa ordem.

c) Ordene todas as notas em ordem crescente utilizando np.sort().

d) Utilize np.argsort() para descobrir os índices que colocariam as notas em ordem crescente.

e) A partir do resultado de argsort(), mostre as notas em ordem decrescente.

```python
notas = np.array([6.5, 8.0, 4.5, 9.0, 7.5, 5.0, 10.0, 7.0])
indices = [1, 3, 6]
print(f'a) Notas dos alunos nas posições 1, 3 e 6: {notas[indices]}')
print(f'b) Nova seleção: {notas[[7, 2, 0, 5]]}')
print(f'c) Notas em ordem crescente: {np.sort(notas)}')
print(f'd) Índices em ordem crescente: {np.argsort(notas)}')
print(f'e) Notas em ordem decrescente: {notas[np.argsort(notas)[::-1]]}')
```

4 ) Uma escola deseja armazenar em um único array o nome, a idade e a nota de seus alunos. Crie o seguinte tipo estruturado:

tipo_aluno = np.dtype([
('nome', 'U20'),
('idade', 'i4'),
('nota', 'f4')
])

Em seguida, crie o array:

alunos = np.array([
('Ana', 19, 8.5),
('Carlos', 21, 7.0),
('Beatriz',20, 9.5),
('Daniel', 18, 6.5)
], dtype=tipo_aluno)

Realize as seguintes operações:

a) Exiba todo o array alunos.

b) Exiba somente os nomes dos alunos.

c) Exiba somente as notas.

d) Mostre apenas os registros dos alunos que possuem nota maior ou igual a 8.0.

e) Ordene o array de alunos pelo campo nota (examinem o parâmetro order do método que usarem).

```python
tipo_aluno = np.dtype([
    ('nome', 'U20'),
    ('idade', 'i4'),
    ('nota', 'f4')
])

alunos = np.array([
    ('Ana', 19, 8.5),
    ('Carlos', 21, 7.0),
    ('Beatriz',20, 9.5),
    ('Daniel', 18, 6.5)
], dtype=tipo_aluno)
```

```python
print('a) Exibindo todo o array alunos:')
print(alunos)
```

```python
print('\nb) Exibindo somente os nomes dos alunos:')
print(alunos['nome'])
```

```python
print('\nc) Exibindo somente as notas:')
print(alunos['nota'])
```

```python
print('\nd) Registros dos alunos com nota maior ou igual a 8.0:')
print(alunos[alunos['nota'] >= 8.0])
```

```python
print('\ne) Array de alunos ordenado pela nota (crescente):')
print(np.sort(alunos, order='nota'))
```

## Resumo Comparativo: Permutação, Combinação Simples e Combinação com Repetição

| Tipo de Cálculo           | Ordem Importa? | Repetição Permitida? | Quando Usar                                              | Fórmula Base (ou Função)                                 |
| :------------------------ | :------------- | :------------------- | :------------------------------------------------------- | :------------------------------------------------------- |
| **1. Permutação**         | SIM            | NÃO                  | Arranjar ou sequenciar itens (a posição faz diferença)  | `P(n, k) = n! / (n - k)!` (`permutacao(n, k)`)         |
| **2. Combinação Simples** | NÃO            | NÃO                  | Selecionar grupos de itens (a ordem não faz diferença)  | `C(n, k) = n! / (k! * (n - k)!)` (`combinacao(n, k)`) |
| **3. Combinação c/ Rep.** | NÃO            | SIM                  | Seleção de itens onde pode haver repetição (a ordem não importa) | `C'(n, k) = C(n + k - 1, k)` (`combinacao_com_repeticao(n, k)`) |

# Desafio 1: Seguência Fibonacci

```python
# Desafio 1: Seguência Fibonacci

def fibonacci(n):
    if n <= 0:
        return []
    fib_sequence = [0, 1]
    while len(fib_sequence) < n:
        fib_sequence.append(fib_sequence[-1] + fib_sequence[-2])
    return fib_sequence

fibonacci(10)
```

```python
# Texte da restrinção de números negativos e 0
fibonacci(0)
```

# Desafio 3: contagem de Caracter


```python
def contar_caracteres(string):
    caracter_count = {}
    string = string.lower() # Converte tudo para minúsculas
    for caracter in string:
        if caracter.isalpha(): # Conta apenas caracteres alfabéticos
            if caracter in caracter_count:
                caracter_count[caracter] += 1
            else:
                caracter_count[caracter] = 1
    return caracter_count
```

```python
contar_caracteres(' Oi, eu sou o bob e moro em são paulo')
```

```python
contar_caracteres('Democracia')
```

```python
contar_caracteres('123456789')
```

```python
contar_caracteres('abcdefghijklmnopqrstuvwxyz babebibibu')
```

# Desafio 4: Número primo


```python
def numero_primo(n):
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
numero_primo(7)
```

```python
numero_primo(10)
```

# Desafio 5: Fatorial


```python
def fatorial(n):
    if n < 0:
        return "Não é possível calcular o fatorial de um número negativo."
    elif n == 0:
        return 1
    else:
        resultado = 1
        for i in range(1, n + 1):
            resultado *= i
        return resultado
```

```python
fatorial(7)
```

```python
fatorial(7)/fatorial(7-3)
```

# Desafio 6: Combinação


```python
def combinacao(n, k):
    if not (isinstance(n, int) and isinstance(k, int) and n >= 0 and k >= 0):
        return "Entrada inválida: n e k devem ser inteiros não negativos."
    if k > n:
        return "O número de itens tomados (k) não pode ser maior que o número total de itens (n)."

    # Usa a função fatorial já definida
    return fatorial(n) / (fatorial(k) * fatorial(n - k))
```

```python
combinacao(7, 3)
```

```python
combinacao("a","b")
```

# Desafio 7: Combinação com repetição

### Exemplo de Uso da Combinação com Repetição

Imagine que você está comprando 3 sorvetes e há 5 sabores disponíveis. Você pode escolher o mesmo sabor várias vezes. Quantas combinações diferentes de 3 sorvetes você pode ter?

*   `n` (número de sabores disponíveis) = 5
*   `k` (número de sorvetes a escolher) = 3

```python
def combinacao_com_repeticao(n, k):
    # Validações de entrada
    if not (isinstance(n, int) and isinstance(k, int) and n >= 0 and k >= 0):
        return "Entrada inválida: n e k devem ser inteiros não negativos."

    # A fórmula para combinação com repetição é C'(n, k) = C(n + k - 1, k)
    # onde C é a combinação simples (sem repetição).
    # Equivalente a: (n + k - 1)! / (k! * (n - 1)!)

    # Podemos reutilizar a função de combinação simples, calculando C(n + k - 1, k)
    # Verificamos se n+k-1 é negativo, embora com n>=0 e k>=0 isso geralmente não aconteça para n,k pequenos.
    # No entanto, o k ainda pode ser maior que o novo 'n' (n+k-1), então a validação da 'combinacao' é suficiente.
    novo_n = n + k - 1
    return combinacao(novo_n, k)
```

```python
n_sabores = 5
k_sorvetes = 3

resultado_repeticao = combinacao_com_repeticao(n_sabores, k_sorvetes)
print(f"Combinações com repetição de {n_sabores} itens tomados {k_sorvetes} a {k_sorvetes}: {resultado_repeticao}")
```

# Desafio 8: Permutação

### Exemplo de Uso da Permutação

Imagine que você tem 4 pessoas (A, B, C, D) e quer escolher 2 delas para ocupar posições específicas (por exemplo, presidente e vice-presidente). A ordem importa aqui.

*   `n` (número total de pessoas) = 4
*   `k` (número de posições a preencher) = 2

```python
def permutacao(n, k):
    # Validações de entrada
    if not (isinstance(n, int) and isinstance(k, int) and n >= 0 and k >= 0):
        return "Entrada inválida: n e k devem ser inteiros não negativos."
    if k > n:
        return "O número de itens tomados (k) não pode ser maior que o número total de itens (n) para permutação simples."

    # A fórmula da permutação de n itens tomados k a k é n! / (n - k)!
    return fatorial(n) / fatorial(n - k)
```

```python
n_pessoas = 4
k_posicoes = 2

resultado_permutacao = permutacao(n_pessoas, k_posicoes)
print(f"Permutações de {n_pessoas} itens tomados {k_posicoes} a {k_posicoes}: {resultado_permutacao}")

# Exemplo de permutação simples (todos os itens):
# Quantas maneiras diferentes você pode organizar 3 livros em uma prateleira?
# n = 3, k = 3
resultado_permutacao_simples = permutacao(3, 3)
print(f"Permutações de 3 itens tomados 3 a 3 (permutação simples): {resultado_permutacao_simples}")
```

# Desafio 9: Calculadora básica ( + - x  / )

```python
def calculadora (x, op, y):
  # Validação para garantir que x e y são números
  if not (isinstance(x, (int, float)) and isinstance(y, (int, float))):
        return "Entrada inválida: x e y devem ser números."

  if(op == '+'):
    return x + y
  elif(op == '-'):
    return x - y
  elif(op == '*'):
    return x * y
  elif(op == '/'):
    if(y != 0):
      return x / y
    else:
      return "Não é possível dividir por zero"
  else:
    return "Operação inválida. Use '+', '-', '*' ou '/'."
```

```python
calculadora(10, '+', 5)
```

```python
calculadora(7, "-", 2)
```

```python
calculadora(3, "*", 4)
```

```python
calculadora( 8, "/", 3)
```

# Desafio 10: Encontrar o Maior Elemento em uma Lista:

```python
def maior_elemento(lista):
    if not lista:
        return None  # Retorna None para lista vazia

    # Inicializa o maior_numero com o primeiro elemento da lista
    maior_numero = lista[0]

    # Itera sobre o restante da lista para encontrar o maior número
    for numero in lista:
        if numero > maior_numero:
            maior_numero = numero

    return maior_numero

# Exemplos de uso:
print(f"Maior elemento em [1, 5, 2, 8, 3]: {maior_elemento([1, 5, 2, 8, 3])}")
print(f"Maior elemento em [-10, -3, -7]: {maior_elemento([-10, -3, -7])}")
print(f"Maior elemento em [5]: {maior_elemento([5])}")
print(f"Maior elemento em []: {maior_elemento([])}")
print(f"Maior elemento em [10.5, 2.3, 7.8, 15.1]: {maior_elemento([10.5, 2.3, 7.8, 15.1])}")
```

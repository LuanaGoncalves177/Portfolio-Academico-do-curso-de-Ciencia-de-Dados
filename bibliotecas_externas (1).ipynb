## Módulos externos

#### Importando um módulo (biblioteca)

````python
import math  # importa o módulo math, para acessar seus métodos, atributos, math.metodo_ou_atributo(...)

math.pow(2, 10)  # quanto é 2 elevado a 10?
````

#### Atribuindo um nome diferente (alias) a um módulo importado

````python
import random as ran

ran.randint(0, 10)  # gera um número aleatório no intervalo [0, 10]
````

#### Importando apenas algumas "coisas" (funções, atributos, constantes, ...) de um módulo
Deixa todas as "coisas" acessíveis sem a necessidade de chamar `modulo.coisa`

````python
from math import sqrt, log2, pi

print(sqrt(2))  # imprime a raíz quadrada de 2
print(log2(16))  # imprime o log de 16 na base 2
print(pi)  # imprime o número pi
````

#### Importando todas as "coisas" (funções, atributos, constantes, ...) de um módulo

````python
from math import *

log10(10)  # log de 10 na base 10
````

Apesar de possível, tal prática **não é indicada**, uma vez que (1) "coisas", p. ex., de módulos diferentes mas de mesmo nome podem ser importadas, e (2) pode não haver necessidade de ter acesso a todas as "coisas" do módulo.

Para ser mais formal, existe uma hierarquia com nomes bem definidos para as "coisas".
A estrutura é:

distribuição -> pacote -> módulo -> submódulo -> função

Por exemplo, usaremos nas próximas aulas a distribuição numpy, que é instalada no ambiente usado, geralmente usando-se o pip:

```python
   pip -m install numpy
```
Em seguida importa-se o pacote de mesmo nome:
```python
   import numpy
```

### Submódulos
Alguns módulos podem possuir **submódulo**:

````python
import numpy

print(f'numpy.random é um {type(numpy.random)}')
````

````python
from numpy import random

random.randint(0, 10)

print(f'numpy.random.randint é um {type(numpy.random.randint)}')
````

# Pickle

````python
import pickle
````

````python
# Dicionário de exemplo com dados
dados = {
    'nome': 'Ana',
    'idade': 28,
    'hobbies': ['ler', 'jogar', 'viajar']
}

# Abre um arquivo no modo de escrita binária ('wb')
with open('dados.pkl', 'wb') as arquivo:
    pickle.dump(dados, arquivo)

print('Dados salvos com sucesso!')
````

O conteúdo do arquivo *dados.pkl* é uma **serialização** do objeto *dados*.
Isto significa que a estrutura (um dicionário, no caso) é salva em um arquivo externo
de modo que possa ser recuperada diretamente depois :

````python
with open("dados.pkl", "rb") as arquivo:
    pessoa = pickle.load(arquivo)
````

```python
# ============================================================
# CÉLULA 1: IMPORTAÇÃO DE BIBLIOTECAS
# ============================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configurações de visualização
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("husl")

print("✅ Bibliotecas importadas com sucesso!")
print(f"Pandas versão: {pd.__version__}")
print(f"NumPy versão: {np.__version__}")
print(f"Matplotlib versão: {plt.matplotlib.__version__}")
```

```python
# ============================================================
# CÉLULA 2: PRIMEIROS PASSOS EM PYTHON
# ============================================================

# Variáveis básicas
nome = "Estudante de Data Science"
semestre = 1
nota_objetivo = 10.0
aprovado = True

print(f"Olá, {nome}!")
print(f"Semestre: {semestre}")
print(f"Objetivo: tirar {nota_objetivo}")
print(f"Aprovado: {aprovado}")

# Operações matemáticas
salario_inicial = 3000
aumento_percentual = 0.10
novo_salario = salario_inicial * (1 + aumento_percentual)

print(f"\nSalário inicial: R$ {salario_inicial:.2f}")
print(f"Aumento: {aumento_percentual*100:.0f}%")
print(f"Novo salário: R$ {novo_salario:.2f}")
```

```python
# ============================================================
# CÉLULA 3: CRIANDO NOSSO PRIMEIRO DATAFRAME
# ============================================================

# Dados de exemplo: funcionários de uma empresa
dados_funcionarios = {
    'Nome': ['Ana Silva', 'Bruno Santos', 'Carla Oliveira', 'Daniel Costa',
             'Eva Pereira', 'Felipe Souza', 'Gabriela Lima', 'Hugo Almeida'],
    'Idade': [25, 30, 28, 35, 42, 27, 33, 29],
    'Departamento': ['TI', 'Marketing', 'TI', 'RH', 'Finanças',
                     'Marketing', 'TI', 'RH'],
    'Salario': [4500, 3800, 5200, 3500, 6000, 4100, 5800, 3700],
    'Anos_Experiencia': [2, 5, 4, 8, 15, 3, 7, 4],
    'Satisfacao': [8.5, 7.2, 9.1, 6.8, 8.9, 7.5, 8.7, 7.0]
}

df_funcionarios = pd.DataFrame(dados_funcionarios)

print("📊 DataFrame de Funcionários:")
print("=" * 60)
print(df_funcionarios)
```

```python
# ============================================================
# CÉLULA 4: EXPLORAÇÃO INICIAL DO DATAFRAME
# ============================================================

print(" EXPLORAÇÃO DO DATAFRAME")
print("=" * 60)

# Dimensões
print(f"\n📐 Dimensões: {df_funcionarios.shape[0]} linhas × {df_funcionarios.shape[1]} colunas")

# Primeiras linhas
print("\n📋 Primeiras 5 linhas:")
print(df_funcionarios.head())

# Informações gerais
print("\nℹ️ Informações sobre as colunas:")
print(df_funcionarios.info())

# Estatísticas descritivas
print("\n📈 Estatísticas Descritivas (variáveis numéricas):")
print(df_funcionarios.describe())
```

```python
# ============================================================
# CÉLULA 5: SELEÇÃO E FILTRAGEM DE DADOS
# ============================================================

print("🎯 SELEÇÃO E FILTRAGEM")
print("=" * 60)

# Selecionar colunas específicas
print("\n📌 Apenas Nome e Salário:")
print(df_funcionarios[['Nome', 'Salario']])

# Filtrar por condição
print("\n Funcionários do departamento de TI:")
ti = df_funcionarios[df_funcionarios['Departamento'] == 'TI']
print(ti)

# Múltiplas condições
print("\n📌 Funcionários com salário > 4000 E satisfação > 8:")
filtro = (df_funcionarios['Salario'] > 4000) & (df_funcionarios['Satisfacao'] > 8)
print(df_funcionarios[filtro])

# Estatísticas por grupo
print("\n📊 Salário médio por departamento:")
print(df_funcionarios.groupby('Departamento')['Salario'].mean())
```

```python
# ============================================================
# CÉLULA 6: CARREGANDO DATASET REAL (IMDB) - CORRIGIDO
# ============================================================

import pandas as pd

# URL atualizada (o dataset original foi movido/removido do GitHub)
url_imdb = 'https://raw.githubusercontent.com/krishna-koly/IMDB_TOP_1000/main/imdb_top_1000.csv'

try:
    # on_bad_lines='skip' evita quebras se houver linhas com vírgulas não tratadas no meio do texto
    # (Nota: se usar Pandas < 1.3.0, troque por error_bad_lines=False)
    df_filmes = pd.read_csv(url_imdb, on_bad_lines='skip')

    print("✅ Dataset IMDB carregado com sucesso!")
    print(f"📊 Total de filmes: {len(df_filmes)}")
    print(f"📋 Colunas: {list(df_filmes.columns)}")
    print("\n🎬 Primeiros 5 filmes:")

    # Exibindo as colunas específicas que você usou
    cols_desejadas = ['Series_Title', 'Released_Year', 'IMDB_Rating', 'Genre']
    print(df_filmes[cols_desejadas].head())

except Exception as e:
    print(f"❌ Erro ao carregar: {e}")
    print("Verifique sua conexão com a internet.")
```

```python
# ============================================================
# CÉLULA 7: ANÁLISE EXPLORATÓRIA DO DATASET IMDB
# ============================================================

if 'df_filmes' in locals():
    print(" ANÁLISE DO DATASET IMDB")
    print("=" * 60)

    # Estatísticas da nota IMDB
    print(f"\n⭐ Nota IMDB - Média: {df_filmes['IMDB_Rating'].mean():.2f}")
    print(f"⭐ Nota IMDB - Mediana: {df_filmes['IMDB_Rating'].median():.2f}")
    print(f"⭐ Nota IMDB - Máxima: {df_filmes['IMDB_Rating'].max()}")
    print(f"⭐ Nota IMDB - Mínima: {df_filmes['IMDB_Rating'].min()}")

    # Top 10 filmes
    print("\n🏆 Top 10 Filmes Mais Bem Avaliados:")
    top10 = df_filmes.nlargest(10, 'IMDB_Rating')[['Series_Title', 'IMDB_Rating', 'Released_Year']]
    print(top10.to_string(index=False))

    # Gêneros mais comuns
    print("\n🎭 Top 10 Gêneros:")
    print(df_filmes['Genre'].value_counts().head(10))
```

```python
# ============================================================
# CÉLULA 8: VISUALIZAÇÕES
# ============================================================

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle(' Análise Exploratória - Dataset IMDB', fontsize=16, fontweight='bold')

if 'df_filmes' in locals():
    # Histograma das notas
    axes[0, 0].hist(df_filmes['IMDB_Rating'], bins=20, color='skyblue', edgecolor='black', alpha=0.7)
    axes[0, 0].set_title('Distribuição das Notas IMDB')
    axes[0, 0].set_xlabel('Nota')
    axes[0, 0].set_ylabel('Frequência')

    # Top 10 gêneros
    top_generos = df_filmes['Genre'].value_counts().head(10)
    axes[0, 1].barh(top_generos.index, top_generos.values, color='coral')
    axes[0, 1].set_title('Top 10 Gêneros de Filmes')
    axes[0, 1].invert_yaxis()

    # Filmes por década
    df_filmes['Decada'] = (df_filmes['Released_Year'] // 10) * 10
    filmes_por_decada = df_filmes['Decada'].value_counts().sort_index()
    axes[1, 0].bar(filmes_por_decada.index, filmes_por_decada.values, color='lightgreen', edgecolor='black')
    axes[1, 0].set_title('Filmes por Década')
    axes[1, 0].set_xlabel('Década')
    axes[1, 0].set_ylabel('Quantidade')
    axes[1, 0].tick_params(axis='x', rotation=45)

    # Boxplot das notas
    axes[1, 1].boxplot(df_filmes['IMDB_Rating'].dropna(), vert=True)
    axes[1, 1].set_title('Boxplot das Notas IMDB')
    axes[1, 1].set_ylabel('Nota')

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# CÉLULA 9: EXERCÍCIO PRÁTICO - DATASET TITANIC
# ============================================================

# Dataset Titanic (disponível no Colab via seaborn)
df_titanic = sns.load_dataset('titanic')

print(" DATASET TITANIC")
print("=" * 60)
print(f"Total de passageiros: {len(df_titanic)}")
print(f"Colunas: {list(df_titanic.columns)}")
print("\n📋 Primeiras linhas:")
print(df_titanic.head())

print("\n❓ Perguntas para investigar:")
print("1. Qual a taxa de sobrevivência geral?")
print("2. Mulheres sobreviveram mais que homens?")
print("3. Passageiros de primeira classe tiveram mais chance?")
print("4. Qual a idade média dos passageiros?")
```

```python
# ============================================================
# CÉLULA 10: RESPOSTAS DO EXERCÍCIO TITANIC
# ============================================================

print("✅ RESPOSTAS - ANÁLISE TITANIC")
print("=" * 60)

# Taxa de sobrevivência
taxa_sobrevivencia = df_titanic['survived'].mean() * 100
print(f"\n1️⃣ Taxa de sobrevivência geral: {taxa_sobrevivencia:.1f}%")

# Sobrevivência por gênero
sobrev_genero = df_titanic.groupby('sex')['survived'].mean() * 100
print(f"\n2️⃣ Sobrevivência por gênero:")
print(f"   - Feminino: {sobrev_genero['female']:.1f}%")
print(f"   - Masculino: {sobrev_genero['male']:.1f}%")

# Sobrevivência por classe
sobrev_classe = df_titanic.groupby('pclass')['survived'].mean() * 100
print(f"\n3️⃣ Sobrevivência por classe:")
for classe, taxa in sobrev_classe.items():
    print(f"   - Classe {classe}: {taxa:.1f}%")

# Idade média
idade_media = df_titanic['age'].mean()
print(f"\n4️ Idade média dos passageiros: {idade_media:.1f} anos")
```

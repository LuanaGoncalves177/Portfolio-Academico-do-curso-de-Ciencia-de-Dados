- Tema: Classificação de espécies de flores Iris

```python
# Carregando o Iris Dataset
import seaborn as sns

df_iris = sns.load_dataset('iris')

# Exibindo as primeiras linhas do dataset
display(df_iris.head())
```

Agora que o dataset está carregado, vamos para o próximo passo: identificar 3 insights e criar 2 visualizações. Que tal começarmos explorando a distribuição das características das flores (sepal_length, sepal_width, petal_length, petal_width)?

### Visualização 1: Pair Plot das Características da Iris
Este gráfico nos ajuda a visualizar a distribuição de cada característica e as relações entre todas as combinações de características, separadas por espécie.

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.pairplot(df_iris, hue='species')
plt.suptitle('Pair Plot das Características da Iris por Espécie', y=1.02)
plt.show()
```

### Visualização 2: Gráfico de Dispersão: Comprimento vs Largura da Pétala por Espécie
Este gráfico é crucial para ver como as espécies de Iris se separam com base nas dimensões de suas pétalas, o que geralmente é um forte indicador.

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(x='petal_length', y='petal_width', hue='species', data=df_iris, s=100, alpha=0.8)
plt.title('Comprimento da Pétala vs Largura da Pétala por Espécie')
plt.xlabel('Comprimento da Pétala (cm)')
plt.ylabel('Largura da Pétala (cm)')
plt.legend(title='Espécie')
plt.grid(True)
plt.show()
```

## 3 Insights Encontrados:
1.  A espécie **Setosa** é claramente distinta das outras duas, tanto no comprimento quanto na largura da pétala, possuindo pétalas significativamente menores.
2.  Há uma sobreposição considerável entre **Versicolor** e **Virginica** em algumas características, mas elas tendem a se separar melhor quando consideramos as dimensões da pétala e do sépal juntas. As Virginicas geralmente têm pétalas e sépalas maiores do que as Versicolor.
3.  As dimensões das pétalas (`petal_length` e `petal_width`) parecem ser os melhores discriminadores para as espécies de Iris, especialmente para separar a Setosa das demais.

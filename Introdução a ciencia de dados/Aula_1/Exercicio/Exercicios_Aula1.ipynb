# Exercício 3: Análise do Dataset Titanic

```python
import seaborn as sns

df_titanic = sns.load_dataset('titanic')

# Responda as perguntas abaixo:
# 1. Qual o percentual de sobreviventes?
# 2. Qual a idade média dos passageiros?
# 3. Quantos passageiros embarcaram em cada porto?
# 4. Qual a tarifa média paga?
# 5. Existem valores ausentes? Em quais colunas?

# Código para começar:
print("1. Taxa de sobrevivência:", df_titanic['survived'].mean() * 100, "%")
print("2. Idade média:", df_titanic['age'].mean())
print("3. Embarque por porto:")
print(df_titanic['embark_town'].value_counts())
print("4. Tarifa média:", df_titanic['fare'].mean())
print("5. Valores ausentes:")
print(df_titanic.isnull().sum())
```

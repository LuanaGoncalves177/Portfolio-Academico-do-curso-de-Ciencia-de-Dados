Exercício 01: Cálculo de consumo médio de Combustível.

O programa deverá armazenar em variáveis:
*   O nome do motorista;
*   A distância percorrida em quilômetros;
*   a quantidade de combustível consumida em litros;
*   o preço fo litro do combustível.

Em seguida, o programa deverá calcular:

*   O consumo médio do veículo em quilômetro por litro;
*   O custo total de combustível utilizado na viagem;
*  O custo médio por quilômetro percorrido.

Ao final, apresente os resultados utilizamdo print() e f-strings.

1. Entrada de variáveis

```python
motorista = " Carlos"
distancia = 420.0
combustivel = 35
preco_litro = 6.00
```

```python
motorista = input("Digite o nome do motorista:")
distancia = float( input("Digite a distância pecorrida:"))
combustivel = int (input("Digite a quantidade de combustível consumida:"))
preco_litro = float (input("Digite o valor do litro de combustível:"))
```

2. Cálculos

```python
consumo_medio = distancia / combustivel
custo_total = combustivel * preco_litro
custo_medio = custo_total / distancia
```

3. Saída de dados

```python
print(f"Nome do motorista:{motorista}")
print(f"Distância percorrida:{distancia}km")
print(f"Quantidade de combustível consumida:{combustivel} litros")
print(f"Preço do litro:{preco_litro}")
```

```python
print(f"Consumo médio: {consumo_medio:.2}")
print(f"Custo total do combustível:{custo_total}")
print(f"Custo médio por quilômetro rodado:{custo_medio:.2}")
```

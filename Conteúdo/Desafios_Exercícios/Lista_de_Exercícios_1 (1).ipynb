Questão: 1. Faça um script python que receba o peso e altura de uma pessoa e compute
seu IMC. Você deve usar uma função pra isso.


```python
# Criação dos inputs e do conversor para a função

peso = float (input("Digite o peso: "))
altura = float (input("Digite a altura: "))
altura, peso
```

```python
# Criação da função que calcula o IMC

def calcular_imc(peso, altura):
  return peso/(altura**2)
```

```python
# Calcular IMC

imc = calcular_imc(peso, altura)
print(f"Seu IMC é: {imc: .5}")
```

Questão: 2. Faça um programa que receba a lista de notas dos alunos de Inteligência
Artificial. Crie uma função que retorne, nesta ordem, a maior nota, menor nota
e nota média.


```python
# 1. Uploard dos dados:

notas_alunos_IA = [4, 8, 5.6, 7, 8, 10, 9, 7, 6, 8.5, 9.5, 9.9]
```

```python
# 2. Criação das variáveis:

maior_nota, menor_nota, media_notas = calculo_de_notas(notas_alunos_IA)
```

```python
# Criação das funções:

def calculo_de_notas(notas_alunos_IA):
  maior_nota = max(notas_alunos_IA)
  menor_nota = min(notas_alunos_IA)
  media_notas = sum(notas_alunos_IA)/ len(notas_alunos_IA)

  return(maior_nota, menor_nota, media_notas)
```

```python
# Exibir resultados

print(f"A maior nota da lista é: {maior_nota:.2f}")
print(f"A menor nota da lista é: {menor_nota:.2f}")
print(f"A média das notas é: {media_notas:.2f}")
```

Questão 3. Utilizando listas, faça um programa que faça 5 perguntas para uma pessoa sobre um crime. As perguntas são:

"Telefonou para a vítima?”
"Esteve no local do crime?”
"Mora perto da vítima?”
"Devia para a vítima?”
"Já trabalhou com a vítima?”

O programa deve no final emitir uma classificação sobre a participação da
pessoa no crime.

Se a pessoa responder positivamente a 2 questões ela deve ser classificada como "Suspeita", entre 3 e 4 como "Cúmplice" e 5 como "Assassino". Caso contrário, ele será classificado como "Inocente".
Utilize o comando input() para ler os valores.

```python
# 1. Criação da função com validação das respostas

def questionario():
  # Criação do Questinário
  print("Responda com 1 para 'Sim' e 0 para 'Não'.")
  respostas = []
  perguntas = [
      "Telefonou para a vítima?",
      "Esteve no local do crime?",
      "Mora perto da vítima?",
      "Devia para a vítima?",
      "Já trabalhou com a vítima?"
  ]

  # Garante que as respostas fornecidas estejam dentro do padrão 0 e 1.

  for i, pergunta_texto in enumerate(perguntas):
    while True:
      try:
        resposta = int(input(f"Questão {i+1}: {pergunta_texto} "))
        if resposta in [0, 1]:
          respostas.append(resposta)
          break
        else:
          print("Por favor, digite apenas 0 ou 1.")
      except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro (0 ou 1).")

  return tuple(respostas)

```

```python
# Soma das resposta da tupla e a classificação do resultado

# Chamando a função e armazenando as respostas
respostas = questionario()

# Contando as respostas positivas
respostas_positivas = sum(respostas)

# Classificando a participação da pessoa
if respostas_positivas == 2:
  classificacao = "Suspeita"
elif 3 <= respostas_positivas <= 4:
  classificacao = "Cúmplice"
elif respostas_positivas == 5:
  classificacao = "Assassino"
else:
  classificacao = "Inocente"

print(f"\nClassificação: {classificacao}")
```

Questão 4: Faça um script que gere uma lista com N números inteiros aleatórios e que conte
quantos números pares e ímpares tal lista possui.


```python

lista_aleatoria = []

def criar_lista_aleatoria(qtd):
  lista_aleatoria = list(range(qtd))
  return(lista_aleatoria)

qtd = int(input("Informe o tamanho da lista aleatória: "))

lista_aleatoria = criar_lista_aleatoria(qtd)

lista_aleatoria
```

```python
def contar_par_e_impar(lista_aleatoria):

  total_pares = 0

  for n in lista_aleatoria:
      if n % 2 == 0:
          total_pares += 1
          total_impares = len(lista_aleatoria) - total_pares

  print("Pares:", total_pares)
  print("Ímpares:", total_impares)
```

```python
# Chamando a função para contar pares e ímpares
contar_par_e_impar(lista_aleatoria)
```

Questão 5: Faça um script python que, dado um texto, indexe suas palavras de acordo com
sua inicial. Ignore vírgulas e a caixa das letras (alta ou baixa)..
P. ex: “o Sapo não lava o pé, mas o Rato roeu a roupa.”
Resultado: {‘r’: [‘rato’, ‘roeu’, ‘roupa’], ‘s’: [‘sapo’], ...}

```python
import string

# Texto de exemplo
texto = "O Sapo não lava o pé, mas o Rato roeu a roupa."

# 1 e 2. Converter para minúsculas e remover pontuações (vírgulas, pontos, etc.)
texto_limpo = texto.lower()
for pontuacao in string.punctuation:
    texto_limpo = texto_limpo.replace(pontuacao, "")

# 3. Separar o texto em palavras
palavras = texto_limpo.split()

# 4. Agrupar no dicionário de acordo com a letra inicial
resultado = {}

for palavra in palavras:
    inicial = palavra[0]  # Pega o primeiro caractere

    # Se a inicial ainda não estiver no dicionário, cria a chave com uma lista vazia
    if inicial not in resultado:
        resultado[inicial] = []

    # Adiciona a palavra na lista correspondente (evitando duplicadas se necessário)
    if palavra not in resultado[inicial]:
        resultado[inicial].append(palavra)

print(resultado)
```

#Desafio 1: Escreva um programa que solicite ao usuário:

1.   O valor do salário base de um vendedor;
2.   A quantidade de itens vendidos por este vendedor.
Dada um valor de comissão, calcule o valor a ser pago a este vendedor.

Para determinar a comissão, use as seguintes regras:
1.   menos de 50 unidades: 5%
2.   menos de 50 até menos de 200 unidades: 10%
3.   200 unidades ou mais: 15%

Mostre o valor calculado.

```python
valor_salario = float(input("Digite o valor do salário base: "))
quantidade_itens = int(input("Digite a quantidade de itens vendidos: "))
```

```python
def comissao(valor_salario, quantidade_itens):
  if quantidade_itens < 50:
    return valor_salario * 0.05
  elif quantidade_itens < 200:
    return valor_salario * 0.10
  else:
    return valor_salario * 0.15
```

```python
valor_comissao = comissao(valor_salario, quantidade_itens)
salario_final = valor_salario + valor_comissao
print(f"Valor da comissão: R${valor_comissao:.2f}")
print(f"Salário base: R${valor_salario:.2f}")
print(f"O valor do salário final é de R${salario_final:.2f}")
```

*   média maior ou igual a 7: "Aprovado"

*   média menor que 5: "Reprovado"

O programa deve continuar executando até que o usuário escolha a opção

```python
agenda = {  "Ana Maria": 999911121,
            "Paulo Lima": 99992548,
            "Cristiane Araúlo": 98887512,
            "Pedro Carsoso": 997412001}
```

```python
# Criando o menu

def menu(): # A função não precisa mais receber 'opcao' como argumento
  while True: # Loop para garantir que uma opção válida seja escolhida
    print("\nMenu:")
    print("1. Adicionar contato")
    print("2. Procurar contato")
    print("3. Remover contato")
    print("4. Listar contatos")
    print("5. Atualizar contato") # Nova opção
    print("6. Sair") # Opção Sair foi movida para 6

    try:
      escolha = int(input("Escolha uma opção: "))
      if 1 <= escolha <= 6: # Condição lógica corrigida para validar a entrada (agora até 6)
        return escolha # Retorna a escolha válida do usuário
      else:
        print("Opção inválida. Por favor, escolha um número entre 1 e 6.")
    except ValueError:
      print("Entrada inválida. Por favor, digite um número inteiro.")
```

```python
# Chamada principal do programa da agenda
while True:
    escolha_usuario = menu() # Chama a função menu para exibir as opções e obter a escolha do usuário

    if escolha_usuario == 1:
      adicionar_contato(agenda)
    elif escolha_usuario == 2:
       procurar_contato(agenda)
    elif escolha_usuario == 3:
        remover_contato(agenda)
    elif escolha_usuario == 4:
        listar_contatos(agenda)
    elif escolha_usuario == 5:
        atualizar_contato(agenda) # Nova chamada para a função de atualização
    elif escolha_usuario == 6: # Opção Sair agora é 6
        print("Saindo da agenda. Até mais!")
        break # Sai do loop, encerrando o programa
```

```python
# Funções do menu:

def adicionar_contato(agenda):
    nome = input("Digite o nome do contato: ")
    telefone = input(f"Digite o telefone de {nome}: ")
    agenda[nome] = telefone
    print(f"Contato '{nome}' adicionado com sucesso!")

def procurar_contato(agenda):
   nome = input("Digite o nome do contato para procurar: ")
   telefone = agenda.get(nome, "Contato não encontrado.")
   print(f"Telefone de {nome}: {telefone}")

def remover_contato(agenda):
    listar_contatos(agenda)
    nome = input("Digite o nome do contato para remover: ")
    if nome in agenda:
        del agenda[nome]
        print(f"Contato '{nome}' removido com sucesso!")
    else:
        print(f"Contato '{nome}' não encontrado na agenda.")

def listar_contatos(agenda):
    if not agenda:
        print("A agenda está vazia.")
    else:
        print("\n--- Lista de Contatos ---")
        for nome_contato, telefone_contato in agenda.items():
            print(f"Nome: {nome_contato}, Telefone: {telefone_contato}")
        print("-------------------------")

def atualizar_contato(agenda):
    listar_contatos(agenda)
    nome = input("Digite o nome do contato que deseja atualizar: ")
    if nome in agenda:
        novo_telefone = input(f"Digite o novo telefone para {nome} (atual: {agenda[nome]}): ")
        agenda[nome] = novo_telefone
        print(f"Contato '{nome}' atualizado com sucesso para {novo_telefone}!")
    else:
        print(f"Contato '{nome}' não encontrado na agenda.")
```

# Desafio 9: Refatoração:

### Organizando as operações do Desafio 9 com um Menu

Para tornar o programa de análise de notas mais interativo e fácil de usar, vamos criar um menu de opções, semelhante ao que fizemos para a agenda. Isso permitirá que você execute as diferentes funcionalidades (mostrar notas, calcular média, alterar nota, adicionar aluno e listar aprovados) de forma dinâmica.

```python
# Função de menu para o Desafio 9
def menu_notas():
    while True:
        print("\n--- Menu de Análise de Notas ---")
        print("1. Mostrar todas as notas e estatísticas")
        print("2. Alterar nota de um aluno")
        print("3. Adicionar novo aluno")
        print("4. Listar alunos aprovados (nota >= 7)")
        print("5. Listar alunos reprovados ")
        print("6. Sair")

        try:
            escolha = int(input("Escolha uma opção: "))
            if 1 <= escolha <= 6:
                return escolha
            else:
                print("Opção inválida. Por favor, escolha um número entre 1 e 6.")
        except ValueError:
            print("Entrada inválida. Por favor, digite um número inteiro.")
```

```python
# Implementação da dictionary comprehension para alunos aprovados
def mostrar_aprovados_comprehension(notas):
    alunos_aprovados = {nome: nota for nome, nota in notas.items() if nota >= 7}
    if alunos_aprovados:
        print("\n--- Alunos Aprovados (Nota >= 7) ---")
        for nome, nota in alunos_aprovados.items():
            print(f"Nome: {nome}, Nota: {nota}")
        print("-------------------------------------")
    else:
        print("Nenhum aluno aprovado com nota maior ou igual a 7 encontrado.")
        print("-------------------------------------------------")

def mostrar_reprovados_comprehension(notas):
    alunos_reprovados = {nome: nota for nome, nota in notas.items() if nota < 7}
    if alunos_reprovados:
        print("\n--- Alunos Reprovados (Nota < 7) ---")
        for nome, nota in alunos_reprovados.items():
            print(f"Nome: {nome}, Nota: {nota}")
```

### Programa Principal de Análise de Notas

```python
# Loop principal para o programa de análise de notas
while True:
    escolha_usuario_notas = menu_notas()

    if escolha_usuario_notas == 1:
        total_notas(notas)
    elif escolha_usuario_notas == 2:
        alterar_nota(notas)
    elif escolha_usuario_notas == 3:
        adicionar_aluno(notas)
    elif escolha_usuario_notas == 4:
        mostrar_aprovados_comprehension(notas)
    elif escolha_usuario_notas == 5:
        mostrar_reprovados_comprehension(notas)
    elif escolha_usuario_notas == 6:
        print("Saindo do programa de análise de notas. Até mais!")
        break
```

produtos = [ {"nome": "Teclado", "preco": 80.0, "quantidade": 5}, {"nome": "Mouse", "preco": 45.0, "quantidade": 8}, {"nome": "Monitor", "preco": 900.0, "quantidade": 3} ]

Crie as seguintes funções: adicionar_produto(...) listar_produtos(...) buscar_produto(...) calcular_valor_estoque(...)

O programa deverá permitir:
* adicionar novos produtos;
* listar todos os produtos;
* procurar um produto pelo nome;
* alterar a quantidade de um produto;
* remover um produto;
* calcular o valor total do estoque.

 O valor de cada item em estoque é calculado por: preço × quantidade Além disso, utilize uma list comprehension para gerar uma lista contendo os nomes dos produtos cujo valor total em estoque seja superior a R$ 500,00. Como desafio adicional, crie, por meio de uma dictionary comprehension, um dicionário no qual a chave seja o nome do produto e o valor seja o valor total daquele produto em estoque.

 Por exemplo: { "Teclado": 400.0, "Mouse": 360.0, "Monitor": 2700.0 }

```python
# Dados iniciais de estoque (lista de dicionários)
produtos = [
    {"nome": "Teclado", "preco": 80.0, "quantidade": 5},
    {"nome": "Mouse", "preco": 45.0, "quantidade": 8},
    {"nome": "Monitor", "preco": 900.0, "quantidade": 3},
    {"nome": "Webcam", "preco": 120.0, "quantidade": 10}
]
```

```python
# Funções para o sistema de controle de estoque

def menu_estoque():
    while True:
        print("\n--- Menu de Controle de Estoque ---")
        print("1. Adicionar produto")
        print("2. Listar todos os produtos")
        print("3. Buscar produto")
        print("4. Alterar quantidade de produto")
        print("5. Remover produto")
        print("6. Calcular valor total do estoque")
        print("7. Produtos com valor total em estoque > R$500")
        print("8. Resumo de valor por produto")
        print("9. Sair")

        try:
            escolha = int(input("Escolha uma opção: "))
            if 1 <= escolha <= 9:
                return escolha
            else:
                print("Opção inválida. Por favor, escolha um número entre 1 e 9.")
        except ValueError:
            print("Entrada inválida. Por favor, digite um número inteiro.")

def adicionar_produto(estoque):
    nome = input("Nome do produto: ")
    preco = float(input("Preço do produto: "))
    quantidade = int(input("Quantidade em estoque: "))
    estoque.append({"nome": nome, "preco": preco, "quantidade": quantidade})
    print(f"Produto '{nome}' adicionado com sucesso!")
    listar_produtos(estoque) # Correção: alterado 'estoques' para 'estoque'

def listar_produtos(estoque):
    if not estoque:
        print("Estoque vazio.")
        return
    print("\n--- Produtos em Estoque ---")
    for produto in estoque:
        print(f"Nome: {produto['nome']}, Preço: R${produto['preco']:.2f}, Quantidade: {produto['quantidade']}")
    print("---------------------------")

def buscar_produto(estoque):
    nome_busca = input("Digite o nome do produto para buscar: ")
    encontrado = False
    for produto in estoque:
        if produto['nome'].lower() == nome_busca.lower():
            print(f"\n--- Produto Encontrado ---")
            print(f"Nome: {produto['nome']}, Preço: R${produto['preco']:.2f}, Quantidade: {produto['quantidade']}")
            print("--------------------------")
            encontrado = True
            break
    if not encontrado:
        print(f"Produto '{nome_busca}' não encontrado no estoque.")
        print("------------------------------------------------")
        resposta = input("Deseja adicionar o produto ao estoque? (sim/não): ")
        if resposta == "sim" or resposta == "s":
            adicionar_produto(estoque)
        else:
          menu_estoque()

def alterar_quantidade(estoque):
    listar_produtos(estoque) # Correção: alterado 'estoques' para 'estoque'
    nome_alterar = input("Digite o nome do produto para alterar a quantidade: ")
    encontrado = False
    for produto in estoque:
        if produto['nome'].lower() == nome_alterar.lower():
            nova_quantidade = int(input(f"Digite a nova quantidade para {produto['nome']} (atual: {produto['quantidade']}): "))
            produto['quantidade'] = nova_quantidade
            print(f"Quantidade de '{produto['nome']}' atualizada para {nova_quantidade}.")
            encontrado = True
            break
    if not encontrado:
        print(f"Produto '{nome_alterar}' não encontrado no estoque.")
        print("------------------------------------------------")
        resposta = input("Deseja adicionar o produto ao estoque? (sim/não): ")
        if resposta == "sim" or resposta == "s":
            adicionar_produto(estoque)

def remover_produto(estoque):
    nome_remover = input("Digite o nome do produto para remover: ")
    produto_removido = False
    for i, produto in enumerate(estoque):
        if produto['nome'].lower() == nome_remover.lower():
            del estoque[i]
            print(f"Produto '{nome_remover}' removido do estoque.")
            produto_removido = True
            break
    if not produto_removido:
        print(f"Produto '{nome_remover}' não encontrado no estoque.")

def calcular_valor_total_estoque(estoque):
    valor_total = sum(p['preco'] * p['quantidade'] for p in estoque)
    print(f"\nValor total do estoque: R${valor_total:.2f}")

def produtos_acima_500(estoque):
    # List comprehension para produtos cujo valor total em estoque seja superior a R$ 500,00
    produtos_caros = [
        p['nome'] for p in estoque if (p['preco'] * p['quantidade']) > 500.0
    ]
    if produtos_caros:
        print("\n--- Produtos com Valor Total em Estoque > R$500 ---")
        for nome_produto in produtos_caros:
            print(f"- {nome_produto}")
        print("-----------------------------------------------------")
    else:
        print("Nenhum produto encontrado com valor total em estoque superior a R$500.")

def resumo_valor_por_produto(estoque):
    # Dictionary comprehension para criar um dicionário de nome do produto e valor total
    valor_por_produto = {
        p['nome']: (p['preco'] * p['quantidade']) for p in estoque
    }
    if valor_por_produto:
        print("\n--- Resumo de Valor por Produto no Estoque ---")
        for nome, valor in valor_por_produto.items():
            print(f"{nome}: R${valor:.2f}")
        print("--------------------------------------------")
    else:
        print("Estoque vazio, sem resumo de valor por produto.")
```

```python
# Loop principal para o programa de controle de estoque
while True:
    escolha_estoque = menu_estoque()

    if escolha_estoque == 1:
        adicionar_produto(produtos)
    elif escolha_estoque == 2:
        listar_produtos(produtos)
    elif escolha_estoque == 3:
        buscar_produto(produtos)
    elif escolha_estoque == 4:
        alterar_quantidade(produtos)
    elif escolha_estoque == 5:
        remover_produto(produtos)
    elif escolha_estoque == 6:
        calcular_valor_total_estoque(produtos)
    elif escolha_estoque == 7:
        produtos_acima_500(produtos)
    elif escolha_estoque == 8:
        resumo_valor_por_produto(produtos)
    elif escolha_estoque == 9:
        print("Saindo do sistema de controle de estoque. Até mais!")
        break
```

# Dicionários

### Entendendo Dicionários em Python

#### 1. Criando um Dicionário
Um dicionário é criado usando chaves `{}` e cada par chave-valor é separado por dois pontos `:`.

```python
# Desafio 7: Criação e manipulação de dicionário de produto

# Solicitar valores ao usuário e armazenar em um dicionário
produto = {
    "nome": input("Digite o nome do produto: "),
    "preco": float(input("Digite o preço do produto: ")),
    "quantidade": int(input("Digite a quantidade do produto em estoque: "))
}

print("\n--- Dados do Produto --- ")
# 1. Mostrar o nome e o preço do produto
print(f"Nome do produto: {produto['nome']}")
print(f"Preço do produto: R${produto['preco']:.2f}")

# 2. Acrescentar a chave 'categoria'
produto["categoria"] = input("Digite a categoria do produto: ")
print(f"Chave 'categoria' adicionada: {produto['categoria']}")

print("\n--- Informações Detalhadas do Dicionário ---")
# 3. Mostrar todas as chaves utilizando keys()
print(f"Chaves do dicionário: {produto.keys()}")

# 4. Mostrar todos os valores utilizando values()
print(f"Valores do dicionário: {produto.values()}")

# 5. Percorrer o dicionário utilizando items() e mostrar cada chave acompanhada de seu valor
print("\nItens do dicionário (chave: valor):")
for chave, valor in produto.items():
    print(f"{chave}: {valor}")

# 6. Calcular o valor total disponível em estoque (preço × quantidade)
valor_total_estoque = produto['preco'] * produto['quantidade']
print(f"\nValor total disponível em estoque para {produto['nome']}: R${valor_total_estoque:.2f}")
```

```python
# Exemplo de criação de um dicionário
minha_agenda = {
    "Alice": "9999-1111",
    "Bob": "9999-2222",
    "Carlos": "9999-3333"
}
print("Minha Agenda:", minha_agenda)
```

#### 2. Acessando Valores
Você pode acessar o valor associado a uma chave usando a chave entre colchetes `[]`.

```python
# Acessando o número de telefone da Alice
print("Telefone da Alice:", minha_agenda["Alice"])

# Tentando acessar uma chave que não existe causará um erro (KeyError)
# print(minha_agenda["Diana"])

# Para evitar erros, você pode usar o método .get()
print("Telefone da Diana (usando get):", minha_agenda.get("Diana", "Contato não encontrado"))
```

#### 3. Adicionando e Modificando Valores
Para adicionar um novo par chave-valor ou modificar um valor existente, basta atribuir um valor a uma chave.

```python
# Adicionando um novo contato
minha_agenda["Diana"] = "9999-4444"
print("Agenda após adicionar Diana:", minha_agenda)

# Modificando o telefone do Bob
minha_agenda["Bob"] = "8888-5555"
print("Agenda após modificar Bob:", minha_agenda)
```

#### 4. Removendo Pares Chave-Valor
Você pode remover um par chave-valor usando a palavra-chave `del` ou o método `.pop()`.

```python
# Removendo Carlos usando del
del minha_agenda["Carlos"]
print("Agenda após remover Carlos com del:", minha_agenda)

# Removendo Alice usando pop() e capturando o valor removido
telefone_alice = minha_agenda.pop("Alice")
print("Agenda após remover Alice com pop():", minha_agenda)
print("Telefone da Alice removido:", telefone_alice)

# O método pop() também permite um valor padrão caso a chave não exista
removido_inexistente = minha_agenda.pop("Fernando", "Contato não encontrado")
print("Tentativa de remover Fernando:", removido_inexistente)
print("Agenda atual:", minha_agenda)
```

#### 5. Métodos Úteis

*   `.keys()`: Retorna uma 'visão' de todas as chaves.
*   `.values()`: Retorna uma 'visão' de todos os valores.
*   `.items()`: Retorna uma 'visão' de todos os pares chave-valor como tuplas.

```python
minha_agenda['Eve'] = '7777-7777' # Adicionando mais um para demonstrar

print("Chaves:", minha_agenda.keys())
print("Valores:", minha_agenda.values())
print("Itens (pares chave-valor):", minha_agenda.items())

print("\nPercorrendo o dicionário:")
for nome, telefone in minha_agenda.items():
    print(f"Nome: {nome}, Telefone: {telefone}")
```

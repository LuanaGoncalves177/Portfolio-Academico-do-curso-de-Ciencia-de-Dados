Exercício 2 — Função para calcular o valor de uma compra

Escreva um script em Python que contenha uma função chamada:
calcular_compra()

A função deverá receber três parâmetros:

*   preco
*   quantidade
*   desconto

quantidade
desconto

O parâmetro desconto deverá possuir valor padrão igual a zero.
A função deverá calcular inicialmente:

subtotal = preço × quantidade

Em seguida, deverá calcular o valor do desconto percentual e o valor final da compra.

A função deverá retornar dois valores:

*  o subtotal da compra;
*  o valor final depois do desconto.

Depois de criar a função, faça três testes:
1. Uma compra de 3 produtos de R$ 50,00, sem informar desconto.

2. Uma compra de 2 produtos de R$ 120,00, com desconto de 10%.

3. Uma compra de 5 produtos de R$ 30,00, com desconto de 15%, utilizando um
argumento nomeado para informar o desconto.

Apresente os resultados utilizando f-strings.

1. Criação de Variáveis

```python
preco, quantidade, desconto = 50, 2, 0.1
```

3. Cálculo do desconto:

```python
def calcular_compra(preco, quantidade, desconto=0):
    subtotal = preco * quantidade
    valor_desconto = subtotal * (desconto / 100)
    valor_final = subtotal - valor_desconto
    return subtotal, valor_final
```

```python
# Teste 1: Uma compra de 3 produtos de R$ 50,00, sem informar desconto.
subtotal1, final1 = calcular_compra(50, 3)
print(f"Teste 1: 3 produtos de R$ 50,00 (sem desconto) - Subtotal: R${subtotal1:.2f}, Final: R${final1:.2f}")

# Teste 2: Uma compra de 2 produtos de R$ 120,00, com desconto de 10%.
subtotal2, final2 = calcular_compra(120, 2, 10)
print(f"Teste 2: 2 produtos de R$ 120,00 (com 10% de desconto) - Subtotal: R${subtotal2:.2f}, Final: R${final2:.2f}")

# Teste 3: Uma compra de 5 produtos de R$ 30,00, com desconto de 15%, utilizando um argumento nomeado para informar o desconto.
subtotal3, final3 = calcular_compra(30, 5, desconto=15)
print(f"Teste 3: 5 produtos de R$ 30,00 (com 15% de desconto) - Subtotal: R${subtotal3:.2f}, Final: R${final3:.2f}")

# Teste 4: Uma compra de 10 produtos de R$ 50,00, com desconto de 25%, utilizando um argumento nomeado para informar o desconto.
subtotal4, final4 = calcular_compra(10, 50, desconto=25)
print(f"Teste 4: 10 produtos de R$ 50,00 (com 25% de desconto) - Subtotal: R${subtotal4:.2f}, Final: R${final4:.2f}")
```

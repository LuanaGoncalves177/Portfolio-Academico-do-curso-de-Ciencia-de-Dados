Diferentemnte das listas em Python, arrays no NumPy só podem conter dados de um mesmo tipo. Caso seja possível, o NumPy irá converter o dado para o tipo mais abrangente (*upcast*)

Podemos também explicitar o tipo de dados do array resultante, usando-se o parâmetro **dtype**

```python
np.array ([1, 2, 3, 4], dtype=np.float32)
```

```python
#Listas aninhadas produzem arrays multidimensionais
np.array([range(i, i + 3) for i in [2, 4, 6]])
```

As listas internas são tratadas como linhas de uma matriz

# Atributos de Arrays

```python
import numpy as np
seq = np.random.default_rng(seed=1298)
```

```python
x1 = seq.integers(10, size=6)      # array unidimensional
x2 = seq.integers(10, size=(3, 4))  # array bidimensional
x3 = seq.integers(10, size=(3,4,5)) # array tridimensional
```

```python
print ("x1: ", x1)
print ("x2: ", x2)
print ("x3: ", x3)
```

```python
print ("x3 ndim:  ", x3.ndim)
print ("x3 shape: ", x3.shape)
print ("x3 size:  ", x3.size)
print ("dtype:    ", x3.dtype)
```

# Fatiamento (*slicing*)

O fatiamento (slicing) funciona igualmente às listas Python

```python
# x[inicio:fim:passo]
x1[:3]  #primeiros três elementos
```

```python
x1[1::2]  # cada segundo elemento do array a partir do índice 1
```

```python
x2
```

```python
x2[:2,:3] #primeiras duas linhas e três colunas
```

```python
x2[::-1,::-1]  #todas as linhas e colunas, em ordem inversa
```

Diferentemente do fatiamento do Python básico, o fatiamento de arrays NumPy retorna como ***view***

Uma ***view*** é uma referência direta ao elemento original e não uma cópia!
Alterações realizadas no subarray
**terão efeito** no array original.

```python
print(x2)
```

```python
x2_sub = x2[:2, :2]
print (x2_sub)
```

```python
x2_sub[0, 0] = 99
print (x2_sub)
```

```python
print (x2)
```

Pode parecer estranho, mas isto permite acessar e processar parte dos conjuntos sem a necessidade de copiar o buffer de dados subjacente

Se houver alguma utilidade em criar um cópia explícita dos dados de um array, a forma mais fácil de fazê-lo é o método copy:

```python
x2_sub_copy = x2[:2, :2].copy()
print(x2_sub_copy)
```

Se modificarmos esse subarray, o array original agora não será alterado:

```python
x2_sub_copy[0,0]=42
print(x2_sub_copy)
```

## Condicionais

São estruturas de comandos existentes em todas as linguagens de programação que permitem que o código examine se uma determinada afirmação é verdadeira ou falsa e, de acordo com este valor, execute diferentes trechos do código.










A afirmação a ser testada pode ser o valor de uma variável ou o resultado de uma expressão lógica, ou mesmo aritmética. Este valor é chamado de **condição**.

Assim, um comando condicional básico em Python tem a forma de
        SE (a condição for verdadeira) faça :
            UMA COISA
        SENÃO faça:
            OUTRA COISA

Como a linguagem tem o inglês como base, o SE é IF e o SENÃO é ELSE.

Se precisarmos encadear várias condições, usaremos outra palavra reservada: ELIF


### Estrutura básica de um IF:
<code>
if condition 01:
    Instruction 01
    Instruction 02
elif (condition 02):
    Instruction 03
    Instruction 04
else:
    Instruction 05
    Instruction 06
<code>
    
Note que as **indentações** definem o _escopo_ de cada opção resultante do IF.

As condições devem resultar um valor boolean (True ou False).

### Operadores de Comparação

| Operação      | Descrição                           |
|---------------|-------------------------------------|
| ``a == b``    | ``a`` igual a ``b``                 |
| ``a < b``     | ``a`` menor do que ``b``            |
| ``a <= b``    | ``a`` menor do que ou igual a ``b`` |

| Operação      | Descrição                           |
|---------------|-------------------------------------|
| ``a != b``    | ``a`` diferente de ``b``            |
| ``a > b``     | ``a`` maior do que ``b``            |
| ``a >= b``    | ``a`` maior do que ou igual a ``b`` |

```python
idade = 17

if idade < 18:
    print('Menor de idade')
else:
    print('Maior de idade')
```

```python
nota = 5.0

if nota < 4.0:
    print('Reprovado direto')
    print("outra linha")
elif nota < 6.0:
    print('Exame Final')
else:
    print('Aprovado')
```

Opcionalmente, podemos usar _parênteses_ nas condições.

```python
nota = 5.0

if (nota < 4.0):
    print('Reprovado direto')
elif (nota < 6.0):
    print('Exame Final')
else:
    print('Aprovado')
```

### Comparando Booleans

```python
aluno_esta_matriculado = True

if aluno_esta_matriculado:
    print('Aluno matriculado')
```

```python
aluno_esta_matriculado = True

if aluno_esta_matriculado is True:
    print('Aluno matriculado')
```

```python
aluno_eh_bruxo = False
#aluno_é_bruxo = True

if not aluno_eh_bruxo:
    print('O aluno não é um bruxo')

if aluno_eh_bruxo is False:
    print('O aluno não é um bruxo')
```

### Combinando condições com `and` e`or`

```python
nota
```

```python
if nota >= 4.0 and nota < 6.0:
    print('Exame')

if (nota >= 4.0) & (nota < 6.0):
    print('Exame')
```

```python
nota = 3.0

if nota < 4.0 or nota >= 6.0:
    print('O aluno não foi para exame')

if (nota < 4.0) | (nota >= 6.0):
    print('O aluno não foi para exame')
```

```python
nota = 3.0
aluno_eh_bruxo = True

if nota < 6.0 and aluno_eh_bruxo:
    print('Temos um mago reprovado')
```

### Armazenando booleans resultantes de condicionais

```python
nota = 7.0
foi_aprovado = (nota >= 6.0)

print(f'Foi aprovado? {foi_aprovado}')
```

```python
nota = 5.0
foi_para_exame = (nota >= 4.0 and nota < 6.0)

print(f'O aluno foi para Exame? {foi_para_exame}')
```

### Condicionais em uma única linha

```python
nota = 3.0

status = 'Aprovado' if nota >= 6.0 else 'Reprovado'

## equivalente a
# if nota >= 6.0:
#     status = 'Aprovado'
# else:
#     status = 'Reprovado'

print(f'Status do aluno: {status}')
```

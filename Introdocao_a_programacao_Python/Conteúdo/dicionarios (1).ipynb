## Dicionários (Dictionaries)
Dicionários são, fundamentalmente, tabelas *hash* : estruturas que ligam _chaves_ a _valores_. <br/>
Muito utilizado quando queremos **acesso rápido** a elementos (valores) associados a uma dada chave.

```python
harry = {'nome': 'Harry Potter', 'idade': 11, 'peso': 35.0}

print(harry)
```

Neste caso, 'nome', 'idade' e 'peso' sãos as **chaves** com os respectivos valores 'Harry Potter', 11, 35.0.

Um dado _valor_ é acessado passando a _chave_ via **colchetes**:

```python
harry['nome']
```

Podemos alterar um valor acessando da mesma maneira:

```python
harry['peso'] = 70

print(harry)
```

Podemos **adicionar** um nome elemento, apenas informando o novo _par de chaves e valor_:

```python
harry['eh_bruxo'] = True
harry['notas'] = [10.0, 5.0, 2.2]

print(harry)
```

Note que usamos **diversos tipos** de _valores_ no mesmo dicionário. Podemos também usar **mais um tipo de chaves**, apesar de, geralmente, isso não ser muito usado.

```python
harry[1] = 'Um'
print(harry)
```

Para copiar um dicionário:

```python
harry_clone = dict(harry)
harry.pop(1, None)  # delete a chave 1 do dicionário harry

print(harry)
print(harry_clone)
```

#### Checando se um dicionário possui uma dada chave

```python
'nome' in harry
```

```python
'profissao' in harry
```

```python
'profissao' not in harry
```

#### For-each em Dicionários

```python
for chave in harry:
    print(f'[{chave}] = {harry[chave]}')
```

```python
harry.keys()  # retorna uma lista com todas as chaves de um dicionário
```

```python
for chave in harry.keys():
    print(f'[{chave}] = {harry[chave]}')
```

```python
harry.values()  # retorna uma lista como todos os valores de um dicionário
```

```python
for valor in harry.values():
    print(f'valor = {valor}')
```

```python
harry.items()  # retorna uma lista de tuplas, onde o primeiro elemento de cada tupla é uma chave e seu segundo elemento é seu respectivo valor indexado
```

```python
for chave, valor in harry.items():
    print(f'[{chave}] = {valor}')
```

#### Dictionary Comprehensions

```python
planetas = ['Mercúrio', 'Vênus', 'Terra', 'Marte', 'Júpiter', 'Saturno', 'Urano', 'Netuno']

iniciais_dos_planetas = {planeta: planeta[0] for planeta in planetas}  # cada planeta será uma chave e sua inicial será seu valor
print(iniciais_dos_planetas)
```

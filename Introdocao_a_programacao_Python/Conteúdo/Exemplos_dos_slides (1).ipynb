```python
import numpy as np
rng = np.random.default_rng(seed=1701)

def calcula_inversos(valores):
    saida = np.empty(len(valores))

    for i in range(len(valores)):
        saida[i] = 1.0 /valores[i]
    return saida

valores = rng.integers(1, 10, size=5)
calcula_inversos(valores)
```

```python
array_grande = rng.integers(1, 100, size=1000000)
%timeit calcula_inversos(array_grande)
```

```python
%timeit (1.0 / array_grande)
```

```python
#calculando usando a função com loop
print (calcula_inversos(valores))
#calculando usando uma operação vetorizada
print (1.0/ valores)
```

Questão 1: Escreva um script em Python que solicite ao usuário uma temperatura em graus Celsius e apresente o valor correspondente em graus Fahrenheit.

A conversão é realizada pela fórmula: [ F = C + 32 ]

O programa deverá possuir uma função chamada celsius_para_fahrenheit que
recebe a temperatura em Celsius como parâmetro e retorna a temperatura
correspondente em Fahrenheit.

Exemplo de execução:

Informe a temperatura em Celsius: 25

Temperatura em Fahrenheit: 77.0



```python
def celsius_para_fahrenheit(temperatura_C):
  temperatura_F = (temperatura_C + 32)
  return temperatura_F

temperatura_C = int(input("Informe a temperatura em Celsius: "))
temperatura_F_resultado = celsius_para_fahrenheit(temperatura_C)
print(f'Temperatura em Fahrenheit: {temperatura_F_resultado}')
```

Questão 2:  Escreva um programa que solicite a idade de uma pessoa e, de acordo com o valor informado, apresente uma das seguintes mensagens:

* Pode votar e pode dirigir
* Pode votar e não pode dirigir
* Não pode votar e não pode dirigir

Crie uma função chamada situacao que receba a idade como parâmetro e
retorne as strings "Pode"e “não pode” e use-a para compor a f-string desejada.
(Assuma que, no Brasil, maiores de 16 anos podem votar mas não podem dirigir
veículos)

Exemplo de execução:

Digite sua idade: 17

Situação: Pode votar e não pode dirigir

```python
def situacao(idade):

  if idade >=18:
    print(f'Situação: Pode votar e pode dirigir')
  elif idade >=16 and idade <18:
    print(f'Situação: Pode votar e não pode dirigir')
  else:
    print(f'Situação: Não pode votar e não pode dirigir')

idade = int(input("Informe a sua idade"))
situacao(idade)
```

Questão 3: Escreva uma função chamada mostra_pares que receba dois números inteiros, inicio e fim.

A função deverá mostrar todos os números pares existentes entre esses dois valores, incluindo os limites quando eles forem pares. O programa
principal deverá solicitar os dois valores ao usuário e chamar a função.

Exemplo de execução:

Valor inicial: 5

Valor final: 14

Números pares:

6

8

10

12

14

```python
valor_inicial = int(input('Valor inicial: '))
valor_final = int(input('Valor Final: '))
def mostra_pares(valor_inicial, valor_final):
  for i in range(valor_inicial, valor_final+1):
    if i % 2 == 0:
      print(i)

mostra_pares(valor_inicial, valor_final)
```

Questão 4: Uma pessoa deseja registrar os valores de várias compras realizadas durante
um dia. Escreva um programa que leia repetidamente o valor de uma compra. O
valor 0 deverá indicar que não existem mais compras a registrar.

Ao final, o programa deverá apresentar:
* a quantidade de compras registradas;
* o valor total das compras;
* o valor médio das compras.

Não considere o valor 0 como uma compra.

Exemplo de execução:
Valor da compra (0 para terminar): 35.50

Valor da compra (0 para terminar): 80.00

Valor da compra (0 para terminar): 24.50

Valor da compra (0 para terminar): 0

Quantidade de compras: 3

Total: R$ 140.00

Média: R$ 46.67

O programa deve funcionar corretamente também quando o usuário digitar 0
logo na primeira leitura.

```python
def registro_de_compras():
  count_compras = 0
  total_valor = 0.0

  while True:
    valor_da_compra = float(input('Valor da compra (0 para terminar): '))

    if valor_da_compra == 0:
      break # Saída do Laço de repetição

    count_compras += 1
    total_valor += valor_da_compra

  # Cálculo da média: Tratando os valores 0 da quantidade de compras
  if count_compras > 0:
    media_valor = total_valor / count_compras
  else:
    media_valor = 0.0 # Se não houver compras a média é 0

  return count_compras, total_valor, media_valor

# Chamada da função
numero_de_compras, valor_das_compras, valor_medio = registro_de_compras()

# Mostra dos valores conforme a formatação requerida
print(f'\nQuantidade de compras: {numero_de_compras}')
print(f'Total: R$ {valor_das_compras:.2f}')
print(f'Média: R$ {valor_medio:.2f}')
```

Questão 5:  Crie uma função chamada maior_numero que:
1. pergunte ao usuário quantos números ele deseja informar;
2. utilize um laço de repetição para realizar a leitura dessa quantidade de
números;
3. determine qual foi o maior número informado;
4. retorne esse valor.
Considere que o usuário deverá informar uma quantidade maior que zero.

O programa principal deverá chamar a função e apresentar o resultado.

Exemplo de execução:

Quantos números deseja informar? 5

Digite o número 1: 12

Digite o número 2: 7

Digite o número 3: 25

Digite o número 4: 18

Digite o número 5: 9

Maior número informado: 25

```python
qtd_numeros = int(input('Quantos Números deseja informar? '))

def maior_numero(qtd_numeros):
  maior_numero = 0
  for i in range(1, qtd_numeros+1):
    lista_numeros = int(input(f'Digite o número {i}: '))
    if lista_numeros > maior_numero:
      maior_numero = lista_numeros
  return maior_numero

resultado = maior_numero(qtd_numeros)
print(f'Maior número informado: {resultado}')
```
